## 1.- CREAR TABLERO

```python
def crear_tablero(tamaño=10):
    tablero = np.full((tamaño, tamaño), "_")
    return tablero
```

## PRUEBAS DISPARAR



```python
def disparar(casilla, tablero):
    fila, columna = casilla
    if tablero[fila, columna] == "O":
        tablero[fila, columna] = "X"
    elif tablero[fila, columna] == "_":
        tablero[fila, columna] = "A"
        tablero["_"] = "A"
    return tablero

print(tablero)
```

## 2.- COLOCAR BARCO

```python
barco_1 = [(0, 1), (1, 1)]
barco_2 = [(1, 3), (1, 4), (1, 5), (1, 6)]

def colocar_barco(barco, tablero):
    for casilla in barco:
        tablero[casilla]="O"
    return tablero

colocar_barco(barco_1, t_barcos_jugador)
colocar_barco(barco_2, t_barcos_rival)
print(t_barcos_rival)
print(t_barcos_jugador)


```

## 3.- CREAR BARCO ESLORA


```python
def crear_barco_eslora(eslora):
    orientacion = np.random.randint(2)
    barco = []
    if orientacion == 0:
            # horizontal
        fila = np.random.randint(10)
        columna = np.random.randint(10 - eslora + 1)
        for columna_actual in range(columna, columna + eslora):
            barco.append((fila, columna_actual))
    else: # vertical
        fila = np.random.randint(10 - eslora + 1)
        columna = np.random.randint(10)
        for fila_actual in range(fila, fila + eslora):
            barco.append((fila_actual, columna))
    return barco
        
```

```python
print(crear_barco_eslora(3))
print(crear_barco_eslora(2))
print(crear_barco_eslora(4))
```

## Crear flota


```python
# 1.- Creo todos los barcos segun la lista esloras_flota
# 2.- Compruebo limites y creo los barcos
# 3.- Compruebo solapamientos y creo barcos
# 4.- Si hay error: crear flota de nuevo
# 5.- Si no hay error: Flota creada

def crear_flota():
    esloras_flota = [4, 3, 3, 2, 2, 2]
    flota = []
    for eslora in esloras_flota:
        flota.append(crear_barco_eslora(eslora))
    for barco in flota:
        for coor in barco:
            for i in coor:
                if i < 0 or i > 9:
                    print("generando flota de nuevo....")
                    return crear_flota()
    ocupadas = []
    for barco in flota:
        for coor in barco:
            if coor in ocupadas:
                print("error")
                return crear_flota()
            else:
                ocupadas.append(coor)
                print("Flota creada")
                return flota



```

```python
def colocar_barcos(tablero):
    flota = crear_flota()
    for barco in flota:
        colocar_barco(barco, tablero)
    return tablero
```


## Crear tablero con la flota



```python
tablero = crear_tablero()
flota = crear_flota()
print("FLOTA: ")
print(flota)

for barco in flota:
    colocar_barco(barco, tablero)

print("TABLERO")
print(tablero)
```

## 5.- DISPARAR BARCOS

```python
def disparo(tablero_rival,tablero_disparos, fila, columna):
    if tablero_rival[fila, columna] == "O":
        tablero_rival[fila, columna]= "X"
        tablero_disparos[fila, columna] = "X"
        print("tocado")
        return True
    else:
        tablero_rival[fila, columna] = "A"
        tablero_disparos[fila, columna] = "A"
        print("agua")
        return False
```
