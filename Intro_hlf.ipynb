## PRUEBAS DISPARAR



```python
def disparar(casilla, tablero):
    fila, columna = casilla
    if tablero[fila, columna] == "O":
        tablero[fila, columna] = "X"
    elif tablero[fila, columna] == "_":
        tablero[fila, columna] = "A"
        tablero["_"] = "A"
    return tablero

```

## PRUEBAS CREAR BARCO ESLORA


```python
def crear_barco_eslora(eslora):
    orientacion = np.random.randint(2)
    barco = []
    if orientacion == 0:
            # horizontal
        fila = np.random.randint(10)
        columna = np.random.randint(10 - eslora + 1)
        for columna_actual in range(columna, columna + eslora):
            barco.append((fila, columna_actual))
    else: # vertical
        fila = np.random.randint(10 - eslora + 1)
        columna = np.random.randint(10)
        for fila_actual in range(fila, fila + eslora):
            barco.append((fila_actual, columna))
    return barco
        
```

```python
print(crear_barco_eslora(3))
print(crear_barco_eslora(2))
print(crear_barco_eslora(4))
```
